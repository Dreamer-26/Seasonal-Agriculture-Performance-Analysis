## Section 1 — Data Loading & Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', palette='Set2', font_scale=1.1)
plt.rcParams['figure.dpi'] = 120

# Load dataset using repository-relative path
df = pd.read_csv('../data/seasonal_agriculture_performance_dataset.csv')
print(f"Dataset shape: {df.shape}")
print(f"\nSeasons present: {df['Season'].unique()}")
df.head()
```

```python
df.info()
```

```python
df.describe().round(2)
```

```python
missing = df.isnull().sum()
print('Missing values per column:')
print(missing[missing > 0] if missing.sum() > 0 else 'No missing values found.')
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
df.dropna(inplace=True)
print(f'\nDataset shape after cleaning: {df.shape}')
```

```python
season_counts = df['Season'].value_counts()
print('Season Counts:\n', season_counts)
fig, ax = plt.subplots(figsize=(6, 4))
season_counts.plot(kind='bar', color=sns.color_palette('Set2', n_colors=3), edgecolor='black', ax=ax)
ax.set_title('Number of Farm Records by Season', fontweight='bold')
ax.set_ylabel('Count')
ax.set_xlabel('Season')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 📋 EDA Summary

The dataset contains farm records spanning three major Indian agricultural seasons — **Kharif, Rabi, and Zaid** — with 28 features covering crop characteristics, environmental conditions, input usage, and financial outcomes.

---
## Section 2 — Yield & Profitability Analysis Across Seasons

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='Season', y='Yield_Tonnes_Ha', order=['Kharif','Rabi','Zaid'], palette='Set2', ax=axes[0], fliersize=3)
axes[0].set_title('Yield (Tonnes/Ha) by Season', fontweight='bold')
axes[0].set_ylabel('Yield (Tonnes/Ha)')
sns.boxplot(data=df, x='Season', y='Profit_INR', order=['Kharif','Rabi','Zaid'], palette='Set2', ax=axes[1], fliersize=3)
axes[1].set_title('Profit (INR) by Season', fontweight='bold')
axes[1].set_ylabel('Profit (INR)')
plt.tight_layout()
plt.show()
yield_profit = df.groupby('Season')[['Yield_Tonnes_Ha','Profit_INR']].agg(['mean','median','std']).round(2)
print('\nYield & Profit Summary by Season:')
print(yield_profit.to_string())
```

```python
season_order = ['Kharif', 'Rabi', 'Zaid']
means = df.groupby('Season')[['Yield_Tonnes_Ha','Profit_INR']].mean().reindex(season_order)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
means['Yield_Tonnes_Ha'].plot(kind='bar', ax=axes[0], color=sns.color_palette('Set2', 3), edgecolor='black')
axes[0].set_title('Mean Yield by Season', fontweight='bold')
axes[0].set_ylabel('Yield (Tonnes/Ha)')
means['Profit_INR'].plot(kind='bar', ax=axes[1], color=sns.color_palette('Set2', 3), edgecolor='black')
axes[1].set_title('Mean Profit by Season', fontweight='bold')
axes[1].set_ylabel('Profit (INR)')
plt.tight_layout()
plt.show()
```

### 📊 Yield & Profitability Interpretation

The boxplots show variation in yield and profit across seasons. Profit distributions have substantial spread, reflecting differences in crops, farm sizes, and market prices. Rabi shows comparatively tighter profit consistency in the analysis.

## Section 3 — Environmental Factors Driving Seasonal Patterns

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.violinplot(data=df, x='Season', y='Rainfall_mm', order=['Kharif','Rabi','Zaid'], palette='Blues', ax=axes[0], inner='quartile')
axes[0].set_title('Rainfall (mm) Distribution by Season', fontweight='bold')
sns.violinplot(data=df, x='Season', y='Avg_Temperature_C', order=['Kharif','Rabi','Zaid'], palette='Reds', ax=axes[1], inner='quartile')
axes[1].set_title('Average Temperature (°C) Distribution by Season', fontweight='bold')
plt.tight_layout()
plt.show()
env_stats = df.groupby('Season')[['Rainfall_mm','Avg_Temperature_C']].agg(['mean','median','std']).round(2)
print('\nEnvironmental Factors Summary:')
print(env_stats.to_string())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=df, x='Rainfall_mm', y='Yield_Tonnes_Ha', hue='Season', hue_order=['Kharif','Rabi','Zaid'], palette='Set2', alpha=0.5, s=25, ax=axes[0])
axes[0].set_title('Rainfall vs Yield by Season', fontweight='bold')
sns.scatterplot(data=df, x='Avg_Temperature_C', y='Yield_Tonnes_Ha', hue='Season', hue_order=['Kharif','Rabi','Zaid'], palette='Set2', alpha=0.5, s=25, ax=axes[1])
axes[1].set_title('Temperature vs Yield by Season', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 🌦️ Environmental Factors Interpretation

**Kharif** is characterized by high rainfall and warm conditions. **Rabi** operates under lower rainfall and cooler temperatures. **Zaid** records higher temperatures and greater dependence on supplemental irrigation.

## Section 4 — Water Efficiency & Disease/Pest Risk Balance

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='Season', y='Water_Efficiency_t_per_1000m3', order=['Kharif','Rabi','Zaid'], palette='YlGnBu', ax=axes[0], fliersize=3)
axes[0].set_title('Water Efficiency (t/1000m³) by Season', fontweight='bold')
sns.boxplot(data=df, x='Season', y='Disease_Pest_Risk_pct', order=['Kharif','Rabi','Zaid'], palette='YlOrRd', ax=axes[1], fliersize=3)
axes[1].set_title('Disease/Pest Risk (%) by Season', fontweight='bold')
plt.tight_layout()
plt.show()
wr_stats = df.groupby('Season')[['Water_Efficiency_t_per_1000m3','Disease_Pest_Risk_pct']].agg(['mean','median','std']).round(2)
print('\nWater Efficiency & Disease Risk Summary:')
print(wr_stats.to_string())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.violinplot(data=df, x='Season', y='Water_Efficiency_t_per_1000m3', order=['Kharif','Rabi','Zaid'], palette='YlGnBu', ax=axes[0], inner='quartile')
sns.violinplot(data=df, x='Season', y='Disease_Pest_Risk_pct', order=['Kharif','Rabi','Zaid'], palette='YlOrRd', ax=axes[1], inner='quartile')
plt.tight_layout()
plt.show()
```

### 💧 Water Efficiency & Disease Risk Interpretation

Kharif presents a high-yield/high-risk trade-off. Rabi achieves a more favourable balance with lower disease risk in the analysis, while Zaid faces water-efficiency challenges under intense summer heat.

---
## Section 5 — Interaction Analysis: Irrigation Method × Season

```python
interaction = df.groupby(['Season', 'Irrigation_Method'])['Water_Efficiency_t_per_1000m3'].mean().unstack()
fig, ax = plt.subplots(figsize=(12, 6))
interaction.plot(kind='bar', ax=ax, edgecolor='black', width=0.75)
ax.set_title('Mean Water Efficiency by Season & Irrigation Method', fontweight='bold')
ax.set_ylabel('Water Efficiency (t/1000m³)')
ax.set_xlabel('Season')
ax.legend(title='Irrigation Method', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print('\nInteraction Table — Mean Water Efficiency (t/1000m³):')
print(interaction.round(3).to_string())
```

```python
interaction_risk = df.groupby(['Season', 'Irrigation_Method'])['Disease_Pest_Risk_pct'].mean().unstack()
fig, ax = plt.subplots(figsize=(12, 6))
interaction_risk.plot(kind='bar', ax=ax, edgecolor='black', width=0.75, colormap='RdYlGn_r')
ax.set_title('Mean Disease/Pest Risk (%) by Season & Irrigation Method', fontweight='bold')
ax.set_ylabel('Disease/Pest Risk (%)')
ax.set_xlabel('Season')
ax.legend(title='Irrigation Method', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print('\nInteraction Table — Mean Disease/Pest Risk (%):')
print(interaction_risk.round(2).to_string())
```

### 🔗 Irrigation × Season Interaction Interpretation

Drip irrigation consistently achieves the highest water efficiency across the three seasons in the analysis, while Flood irrigation records the lowest. Rainfed performs reasonably well in Kharif but is less suitable for Zaid's water-stressed conditions.
